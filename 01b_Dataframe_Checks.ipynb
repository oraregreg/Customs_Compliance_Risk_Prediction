```python
import pandas as pd
df = pd.read_csv("customs_clearance_data.csv")
df.shape
```

```python
df["slow_clearance"].value_counts()
```

```python
# To see the number of rows that have null values
df.isnull().sum()
```

```python
df["hs_code"].str.len().value_counts()
```

```python
#  checking if the missing COO rows has anything to do with
# slow clearence

df["country_origin"].isnull()                                   # True/False per row
df.groupby(df["country_origin"].isnull())["slow_clearance"].value_counts(normalize=True)
```

```python
#  checking if the missing HS code rows has anything to do with
# slow clearence
df["hs_code"].isnull().sum()                                      # how many are missing
df.groupby(df["hs_code"].isnull())["slow_clearance"].value_counts(normalize=True)
```

```python
# checking to see that the length of the HS codes is okay
df["hs_code"].str.len().value_counts()
```
